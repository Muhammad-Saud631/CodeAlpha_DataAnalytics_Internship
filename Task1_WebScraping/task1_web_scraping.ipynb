```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "https://books.toscrape.com/"

response = requests.get(url)

print(response.status_code)
```

```python
soup = BeautifulSoup(response.text, "html.parser")

print(soup.title.text)
```

```python
books = soup.find_all("article", class_="product_pod")

print("Total Books:", len(books))
```

```python
first_book = books[0]

title = first_book.h3.a["title"]

print(title)
```

```python
price = first_book.find("p", class_="price_color").text

print(price)
```

```python
rating = first_book.p["class"][1]

print(rating)
```

```python
book_names = []
prices = []
ratings = []

for book in books:
    
    # Book Name
    title = book.h3.a["title"]
    book_names.append(title)
    
    # Price
    price = book.find("p", class_="price_color").text
    prices.append(price)
    
    # Rating
    rating = book.p["class"][1]
    ratings.append(rating)

print("Data Extracted Successfully")
```

```python
df = pd.DataFrame({
    "Book Name": book_names,
    "Price": prices,
    "Rating": ratings
})

df
```

```python
df.to_csv("books_dataset.csv", index=False)

print("CSV File Saved Successfully")
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

book_names = []
prices = []
ratings = []

# Loop for 2 pages
for page in range(1, 3):

    if page == 1:
        url = "https://books.toscrape.com/"
    else:
        url = f"https://books.toscrape.com/catalogue/page-{page}.html"

    response = requests.get(url)

    soup = BeautifulSoup(response.text, "html.parser")

    books = soup.find_all("article", class_="product_pod")

    for book in books:

        # Book Title
        title = book.h3.a["title"]
        book_names.append(title)

        # Price
        price = book.find("p", class_="price_color").text
        prices.append(price)

        # Rating
        rating = book.p["class"][1]
        ratings.append(rating)

# Create Dataset
df = pd.DataFrame({
    "Book Name": book_names,
    "Price": prices,
    "Rating": ratings
})

# Save CSV
df.to_csv("Dataset_of_books.csv", index=False)

print(df.head())

print("\nTotal Books Scraped:", len(df))
print("Dataset Saved Successfully")
```
