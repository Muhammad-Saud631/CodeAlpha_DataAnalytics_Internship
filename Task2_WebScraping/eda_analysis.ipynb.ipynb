```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Dataset_of_books.csv")

df.head()
```

# Questions for Analysis

1. Which book rating is most common?
2. What is the average price of books?
3. Are expensive books highly rated?
4. Are there any missing values?
5. What patterns can we identify from the dataset?

```python
print(df.shape)
```

```python
print(df.dtypes)
```

```python
print(df.isnull().sum())
```

```python
df["Price"] = df["Price"].str.encode("ascii", "ignore").str.decode("ascii")

df["Price"] = df["Price"].str.replace("£", "", regex=False)

df["Price"] = df["Price"].astype(float)

print(df.head())
```

```python
df.describe()
```

```python
df["Rating"].value_counts()
```

```python
df["Rating"].value_counts().plot(kind="bar")

plt.title("Book Ratings Distribution")
plt.xlabel("Ratings")
plt.ylabel("Count")

plt.show()
```

```python
average_price = df["Price"].mean()

print("Average Price:", average_price)
```

```python
most_expensive = df[df["Price"] == df["Price"].max()]

print(most_expensive)
```
